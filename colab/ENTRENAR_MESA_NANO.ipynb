```python
from ultralytics import YOLO
m = YOLO('yolo11n-seg.pt')  # seg, NO el yolo11n.pt normal
m.train(data='synth_lote1/data_colab.yaml', epochs=60, imgsz=640, batch=16, name='mesa_nano', task='segment')
```

```python
m.val()
```

```python
# Prueba con 1 sintética del lote (luego sube 1 foto REAL tuya y cambia el nombre)
m.predict('synth_lote1/images/syn_0000.jpg', save=True, conf=0.35)
from IPython.display import Image
Image('runs/segment/mesa_nano/syn_00000.jpg', width=480)
```
