```python
from ultralytics import YOLO
m = YOLO('yolo11n-seg.pt')  # seg, NO el yolo11n.pt normal
m.train(data='synth_lote2/data_colab.yaml', epochs=60, imgsz=640, batch=32, name='mesa_v2', task='segment')
```

```python
m.val()
```
